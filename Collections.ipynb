```python
#LiSt functions
colors = ["red","green","black","blue"]
print(len(colors))
colors.append("gold")
print(colors)
0#replace an element
colors[1] = "white"
print(colors)
colors.insert(2,"pink")
print(colors)
#remove an element
colors.remove("black")
print(colors)
#remove the last element
colors.pop()
print(colors)
```

```python
#Looping
colors = ["red","green","black","blue"]
#Using index
for i in range(len(colors)):
    print(i, colors[i])
#Directly
for color in colors:
    print(color,end=" ")
```

```python
marks_sem1 = [80, 90, 70, 60, 50]
print("Max:",max(marks_sem1))
print("Min:",min(marks_sem1))
marks_sem1.sort()
print(marks_sem1)
marks_sem1.reverse()
print(marks_sem1)

```

```python
marks_sem1 = [80, 90, 70, 60, 50]
marks_sem2 = [180, 190, 170, 160, 150]
# final_marks = marks_sem1 + marks_sem2
# print(final_marks)
#store both sem marks in marks_sem1 list
marks_sem1.extend(marks_sem2)
print(marks_sem1)
```

```python
#List coomprehensions
#rule -> new_list = [expression for item in iterable if condition]
n = 10
even = [i for i in range(n+1) if i%2 == 0]
print(even)

x = 5
sqr =[i**2 for i in range(1,x+1)]
print(sqr)
```
