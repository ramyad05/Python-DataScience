NumPy array and basics

```python
import numpy as np
```

 Array Creation

```python
# 1D array creation
arr_1d = np.array([10,20,30,40])
print(arr_1d, type(arr_1d))
arr_2d = np.array([
                    [10,20,30],
                    [40,50,60]
                 ])
print(arr_2d, type(arr_2d))
arr = np.array([
                    [10,20,45.5],
                    [40,50,60]
                 ])
print(arr, type(arr))
```
