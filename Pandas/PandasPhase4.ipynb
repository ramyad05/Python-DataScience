### GroupBy, MergeOrJoin

```python
import pandas as pd
```

```python
exp = pd.read_excel("subdata_pandas/expense (1).xlsx")
exp
```

```python
# Find the total expense of each category

exp.groupby("Category")['Amount'].sum() #M1
exp.groupby("Category")['Amount'].agg(['sum']) #M2
```

```python
# give a complete expense summary for each category

exp.groupby("Category")['Amount'].agg(['sum','count','mean','max','min'])
```

```python
# How much money was paid by each payment method

exp.groupby("Payment Mode")['Amount'].agg(['sum'])
```

```python
# How many transactions were made by each payment mode
exp.groupby("Payment Mode")['Amount'].agg(['count'])
```

```python
# For each payment mode, calculate the total amount spent across categories
exp.groupby(['Payment Mode','Category'])['Amount'].agg(['sum','count'])
```

```python
# Which category have the highest total expenses

exp.groupby("Category")['Amount'].sum().sort_values(ascending=False)
```

### MergeOrJoin

```python
customer = pd.read_csv('subdata_pandas/customer.csv')
customer
payment = pd.read_csv('subdata_pandas/payment.csv')
payment
```

```python
#Inner Join

pd.merge(left=customer, right=payment,on='customerid',how='inner')
```

```python
#Left Join

pd.merge(left=customer, right=payment,on='customerid',how='left')
```

```python
#Right Join

pd.merge(left=customer, right=payment,on='customerid',how='right')
```

```python
#Cross Join

pd.merge(left=customer, right=payment, how='cross')
```
