### Dealing with Duplicate values, missing data

```python
import pandas as pd
import numpy as np
```

### Duplicate Values

```python
cdata = pd.read_csv("subdata_pandas/company.csv")
cdata
```

```python
#Overall duplicate
cdata.duplicated() #Boolean
cdata[cdata.duplicated()]  #Select row

#Based on
cdata['EEID'].duplicated()  #Boolean
cdata[cdata['EEID'].duplicated()]

cdata.drop_duplicates()
cdata['EEID'].drop_duplicates()
```

### Missing Data

```python
cdata = pd.read_csv("subdata_pandas/company.csv")
cdata
```

```python
#Check null
cdata.isnull()
#Employee
edata = pd.read_csv("subdata_pandas/employee.csv")
edata.isnull().sum()

edata['salary'].isnull().sum()
#Eliminate Nan rows
cdata.dropna()
```

```python
#Replace NaN with any value
cdata.replace(np.nan,"Insufficient")

#Replace NaN with 30000 in salary
cdata['salary'].replace(np.nan,3000)

#Replace NaN salary with avg
mean = np.mean(cdata['salary'])
mean

#M1
cdata['salary'].replace(np.nan,mean)

#M2
cdata['salary']= cdata['salary'].fillna(mean)
cdata
```

```python
cdata = pd.read_csv("subdata_pandas/company.csv")
cdata

# Forword fill (Previous row)
cdata.ffill()

# Backword fill (Next row)
cdata.bfill()

#Save it
cdata.bfill(inplace=True)
cdata

cdata['salary']=cdata['salary'].bfill()
cdata
```
