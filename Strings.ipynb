String slicing and indexing

```python
#indexing
str = "DARKCHOCOLATE"
print("org:",str)
#Idx +ve
#Retrieve the fifth element
print("5th",str[5])
#last element
print("Last", str[len(str)-1])
#Idx -ve
#Print the last element
print("Last",str[-1])
#Print the first element
print("first",str[-len(str)])
```

```python
#Slicing - str[start:end:step]
# print(str[0:4])
# print(str[4:])
# print(str[0:len(str):2])
# print(str[:])
# print(str[12:3:-1])
print(str[::-1])
print(str[9:])
```

String functions

```python
s ="raju told me that he is a coder"
#Concatenation
print(s+ ", it's real")
print("Length",len(s))
#is stirng ends with substring
print(s.endswith("er"))
#replace old to new
print(s.replace("he is","I am"))
#first index of substring 
print(s.find("a"))
#count the frequency of suubstring
print(s.count("e"))
#Capitalize the first letter of the string
print(s.capitalize())
#Uppercase
print(s.upper())
#Lowercase
print(s.lower())
```
