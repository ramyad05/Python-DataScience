### Measure of Variability

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### Range

```python
titanic_df = pd.read_csv('titanic.csv')
titanic_df.head(10)
```

```python
#Range of ticket fare
max_fare = np.max(titanic_df['Fare'])
min_fare = np.min(titanic_df['Fare'])
range_Fare = max_fare - min_fare
max_fare, min_fare, range_Fare
```

```python
#Range of Age
max_age = np.max(titanic_df['Age'])
min_age = np.min(titanic_df['Age'])
range_Age = max_age - min_age
max_age, min_age, range_Age
```

### Mean Absolute Diviation

```python
IT_Department_A = {
    "Employee ID": ["E101", "E102", "E103", "E104", "E105", "E106"],
    "Employee Name": ["Rohan", "Kiran", "Aditya", "Sneha", "Pooja", "Rahul"],
    "Project": [75, 65, 73, 68, 72, 67]
}

IT_Department_B= {
    "Employee ID": ["E201", "E202", "E203", "E204", "E205", "E206"],
    "Employee Name": ["Arjun", "Deepak", "Kavya", "Megha", "Sanjay", "Varun"],
    "Project": [90, 47, 93, 47, 96, 51]
}

deptA_df = pd.DataFrame(IT_Department_A)
deptB_df = pd.DataFrame(IT_Department_B)
```

```python
plt.scatter([i for i in range(0, len(deptA_df['Project']))],deptA_df['Project'],
            s=100, c='red', label='Dept A')
plt.scatter([i for i in range(0, len(deptB_df['Project']))],deptB_df['Project'],
            s=100, c='green', label='Dept B')
meanVal = np.mean(deptA_df['Project'])
plt.axhline(y=meanVal, color='orange',linestyle='--',label='Mean')
plt.ylim(30,110)
plt.legend()
plt.show()
```

```python
#Find MAD IT_Department_A
x_i = deptA_df['Project']
x_bar_mean = np.mean(deptA_df['Project'])
n = len(deptA_df['Project'])
deptA_MAD = np.sum(np.abs(x_i - x_bar_mean)) / n
print("DeptA MAD:", deptA_MAD)
```

```python
#Find MAD IT_Department_B
x_i = deptB_df['Project']
x_bar_mean = np.mean(deptB_df['Project'])
n = len(deptB_df['Project'])
deptB_MAD = np.sum(np.abs(x_i - x_bar_mean)) / n
print("DeptB MAD:", deptB_MAD)
```
