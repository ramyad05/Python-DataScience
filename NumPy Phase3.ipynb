### NumPy Functions

```python
import numpy as np
```

### Aggregate function

```python
arr = np.array([10,20,30,40,50])
print("Sum:",np.sum(arr))
print("Max:",np.max(arr))
print("Min:",np.min(arr))
print("Count:",arr.size)
print("Commulitive Sum:",np.cumsum(arr))
print("Commulitive Product:",np.cumprod(arr))
```

### Stastical function

```python
import statistics as st
sales =np.array([200,180,150,250,270,300,250,230,500])
print("Mean:",np.mean(sales))
print("Median:",np.median(sales))
print("Mode:",st.mode(sales))
print("Standard Deviation:",np.std(sales))
```

```python
price = np.array([120,180,200,210,220,300])
sales = np.array([15,13,10,12,9,5])
print(np.corrcoef([price,sales]))
```
