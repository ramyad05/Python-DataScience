### Introduction to MatplotLib

```python
import matplotlib.pyplot as plt
import pandas as pd
```

### Bar Plot

```python
course_enroll = {
"courses": ['Java', 'Python','C', 'C++'],
"enrolled": [25,15,18,10]
}
course_df = pd.DataFrame(course_enroll)

#Create plot
plt.Figure(figsize=(9,5))
bars = plt.bar(course_df["courses"],course_df["enrolled"],
        color = ["#4745A1","#A1A147","#A14747","#47A147"],
        edgecolor = "black",width = 0.6
        )

# Creating title and labels
plt.title("Course Enrollment Student Overview",fontsize = 15, fontweight = 'bold', pad=5)
plt.xlabel("Course Details",fontsize = 12)
plt.ylabel("Course Enrolled Students",fontsize = 12)
plt.bar_label(bars, padding = 2,fontweight = 'bold',fontsize = 12)

#Improve
plt.ylim(0,30)
plt.grid(axis='y',linestyle="--",alpha=0.3)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
    
```

### Line plot

```python
weekly_sales = {
    "days":['Mon','Tue','Wed','Thu','Fri','Sat','Sun'],
    'first_week':[35000,33000,38000,32000,39000,42000,45000],
    'second_week':[32000,35000,42000,37000,34000,43000,48000],
}
sales = pd.DataFrame(weekly_sales)
plt.Figure(figsize=(9,5))
#--------First Week Sales---------
plt.plot(sales["days"],sales['first_week'],
         color = "red",ls="--",marker = "o",
         label = "First Week Sales"
         )
#--------Second Week Sales---------
plt.plot(sales["days"],sales['second_week'],
         color = "blue",ls="--",marker = "s",
         label = "Second Week Sales"
         )

#Create title label
plt.title("Shop Weekly Sales Overview",fontsize = 15, fontweight = 'bold')
plt.xlabel("Days of Week",fontsize = 12)
plt.ylabel("Sales amount per Day",fontsize = 12)
plt.legend()
plt.tight_layout()
plt.show()
```

### Scatter Plot

```python
student_age = {
    "age": [20, 22, 21, 20, 21, 22, 23, 22, 20, 24, 22,20, 22, 21, 20, 21, 22, 23, 22, 20, 24, 22],
    "marks":[490,465,325,400,450,380,500,470,420,360,330,490,265,325,300,450,380,500,270,420,460,330]
}
std_df = pd.DataFrame(student_age)
plt.Figure(figsize=(9,5))
plt.scatter(std_df['marks'],std_df['age'],
            color = "red",marker = "o",
            s=100, alpha = 0.3,
            label = "age/marks"
            )
#Create title and label
plt.title("Based on age check marks distributuion",fontsize = 15, fontweight = 'bold')
plt.xlabel("Marks",fontsize = 12)
plt.ylabel("Student Age",fontsize = 12)
plt.legend()
plt.show()
```

### Pie Chart

```python
weekly_sales ={
    'Brand':["Oneplus","Apple","Oppo","Vivo","Samsung"],
    'Sales':[23,10,19,32,25]
}
sales_df = pd.DataFrame(weekly_sales)
plt.pie(
    sales_df['Sales'],
    labels = sales_df['Brand'],
    explode = (0,0,0.1,0,0),
    shadow = True,
    autopct = "%.2f%%",
    pctdistance = 0.7,
    startangle = 30
)
plt.title("Weekly Sales Overview",fontweight = 'bold')
plt.legend(loc="center right",
           bbox_to_anchor=(1.3,0.8))       
plt.show()
```

```python
data = pd.read_excel("expense (1).xlsx")
exp = pd.DataFrame(data)
exp_group = exp.groupby("Category")['Amount'].sum()
plt.pie(
    exp_group.values, labels = exp_group.index,
    autopct = "%.2f%%",explode = (0,0.1,0,0,0),
    shadow = True
)
plt.show()
```

### Histogram

```python
temp_celcius = [28,29,30,31,31,30,29,28,27,26,25,24,24,24]
plt.hist(
    temp_celcius, 
    bins = 5,
    color = 'green',
    alpha = 0.5,
    edgecolor = 'black'
)
plt.title("Distribution of Today's Hourly Temperature", fontweight='bold',pad=5)
plt.xlabel("Temperature",fontsize=12)
plt.ylabel("Frequency",fontsize=12)
plt.grid(axis='y',linestyle='--',alpha=0.3)
plt.show()
```

### Stack Plot

```python
# Weekly Sales Data (in units)
weekly_sales = {
    "Days": ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"],
    "Week1": [5, 10, 30, 20, 35, 60, 80],
    "Week2": [30, 60, 30, 70, 80, 50, 30],
    "Week3": [10, 40, 50, 23, 69, 82, 43]
}
sales = pd. DataFrame(weekly_sales)
plt.stackplot(
    sales['Days'],sales['Week1'],sales['Week2'],sales['Week3'],
    labels = ["Week1","Week2","Week3"]
)
plt.legend()
plt.show()
```
