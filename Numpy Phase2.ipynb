### NumPy Array Operation

```python
import numpy as np
```

```python
arr = np.array([10,20,30,40,50])
print("Basic Slicing:", arr[:4])
print("Slicing with step:", arr[0:5:2])
print("Negative index:", arr[-1])

```

```python
arr_2d = np.array([[1,2,3,10],
                   [4,5,6,20],
                   [7,8,9,30]])
print("Specific element:", arr_2d[1,2])
print("Entire row:\n", arr_2d[1:,2:])
#Last two rows
print("Last two rows:\n", arr_2d[1:,0:])
print("Entire one col:",arr_2d[:,1])
print("Entire col:",arr_2d[:,1:])

sales = np.array([[1,2000,3000,1000],       #Res1
                  [2,5000,6000,2000],       #Res2
                  [3,8000,9000,3000]])      #Res3
print("Last 2 year sales", sales[1,2:])
```

### Sorting

```python
arr_1d = np.array([3,5,2,9,5,6,1,7])
print("Sorting 1d array", sorted(arr_1d))
print("Sorting 1d array", np.sort(arr_1d))

arr_2d = np.array([[2,1],
                   [3,2],
                   [1,3]])
print("Sorting 2d array", np.sort(arr_2d))
# axis=0 for col
print("Sorting 2d array", np.sort(arr_2d,axis=0))
# axis=1 for row
print("Sorting 2d array", np.sort(arr_2d,axis=1))


```

### Filter

```python
number = np.array([3,5,2,9,5,6,1,7])
even_number = number[number % 2 == 0]
print("Even Number:",even_number)
```

### Filter with mask

```python
marks = np.array([67,34,10,23,50,97,54])
mask_pass = marks > 35
print("More than 35(Mask): ",mask_pass)
```

### Index

```python
sales = np.array([25000,12000,65000,80000])
idc = [1,3]
res = sales[idc]
print("Find value in index:",res)
```
