Python if-else


```python
age = int(input("Enter the age: "))
if age >= 18:
    print("Allowed for toxic")
else: 
    print("Not allowed for toxic")
```

Python if-elif-else

```python
age = int(input("Enter age: "))
if age <= 3:
    print("Child")
elif age >= 4 and age <= 12:
    print("Kid")
elif age >= 13 and age <= 18:
    print("Teenage")
elif age >= 19 and age <= 60:
    print("Adult")
else:
    print("Thatha")
```

Short Hand

```python
num = 25
print("Even") if num % 2 == 0 else print("Odd")
```
