Python tuples

```python
#Types
#Type1: Using Parenthese
nums = (1, 2, 3, 4, 5)
colors = ("red", "green", "blue")
mixed = ("Raj", 23, 54.8, True)
nested = (23, (2,4), [2,5,6])

#Type2:Without Parenthese
fruits = "apple", "banana", "cherry"
print(type(fruits))

#Type3: Using tuple() constructor
values = tuple((1,2,3,4,5,6))
print(values,tuple(values))
list_item = [10,20,30]
list_tup = tuple(list_item)

#Type4: Single value tuple(using comma)
num = (55,)
print(num,type(num))
```

Tuples Immutable

```python
colors = ("red", "blue", "green")
# colors[1] = "pink" #Does not support
print(colors)
#if u want to change , convert to list
list_colors = list(colors)
list_colors[1] = "pink"
colors = tuple(list_colors)
print(colors,type(colors))

```

Tuple Packing & Unpacking

```python
v1 = "Raj"
v2 = 23
v3 = "CSE"
#Packing
student = v1, v2, v3
#Unpacking
name, age,dept = student
print(name)
print(age)
print(dept)
```

Tuple methods and functions


```python
nums = (10,30,50,20,40,15,50)
print(max(nums))
print(min(nums))
sort_nums = sorted(nums)
print(sort_nums)
print("Count",nums.count(50))
print("Index",nums.index(30))
```
