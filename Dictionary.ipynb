Dictionary functions and methods

```python
student = {
    'id':"S2601",
    'name':"Ramya",
    'age':20,
    'dept':"CSEB"
}

#Retrieve a single value using key
get_value = student.get('age')
print(get_value)

#Check all the keys
dict_keys = student.keys()
print(dict_keys)

#Retrive all the values
dict_values =  student.values()
print(dict_values)

#Update
update_age = student.update({'age':19})
print(student)

#Set default
dep_val = student.setdefault("dept","CS")
print(dep_val)
new_loc = student.setdefault("place","Udupi")
print(student)

#copy
c = student.copy()
print(c)

#pop- remove key-value based on the key
student.pop("dept")
print(student)

#popitem
student.popitem()
print(student)

print(student.items())
```

Dict Iteration

```python
student = {
    'id':"S2601",
    'name':"Ramya",
    'age':20,
    'dept':"CSEB"
}

#Itr over keys
for k in student:
    print(k)
for v in student.values():
    print(v)
for k,v in student.items():
    print(k,":",v)
```

 Write a script to generate a lookup dictionary where keys represent risk tiers(1 to 15) & values are the square of each tier number.

```python
dict_square = {}
for i in range(1,16):
    dict_square.setdefault(i,i**2)
print(dict_square)
```

You manage a retail sales register where customer checkout totals arrive throughout the day in a dictionary format: {"Customer Name":amount_paid}. Write a pyhton program to sort this sales dictionary by transaction amout in descending order and identify the top-paying customer.

```python
# Sales dictionary
sales = {
    "Rahul": 2500,
    "Anu": 4500,
    "Priya": 3200,
    "Kiran": 1800,
    "Sneha": 5200
}

# Sort the dictionary by transaction amount in descending order
sorted_sales = dict(sorted(sales.items(), key=lambda item: item[1], reverse=True))

# Display sorted sales
print("Sales in descending order:")
for customer, amount in sorted_sales.items():
    print(customer, ":", amount)

# Identify the top-paying customer
top_customer = max(sales, key=sales.get)

print("\nTop-paying customer:")
print(top_customer, "paid", sales[top_customer])
```
